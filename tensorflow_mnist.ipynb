```python
x_train = x_train.reshape(-1,28*28)
x_test = x_test.reshape(-1,28*28)
```

```python
print(x_train[0])
```

```python
#ANN
model = keras.Sequential([
    keras.layers.Dense(3000,activation='relu',input_shape=(784,)),  #we are just flatening the data in single row insted of 28x28
    keras.layers.Dense(1000,activation='relu'),  #passing through every layer is just geting prob. of the value..
    keras.layers.Dense(500,activation='relu'),
    keras.layers.Dense(10,activation='softmax')])
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
model.fit(x_train,y_train,epochs=2)
```

```python
model.evaluate(x_test,y_test)
```
